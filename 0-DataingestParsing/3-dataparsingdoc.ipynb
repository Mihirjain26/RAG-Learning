### Word Document processing

```python
from langchain_community.document_loaders import Docx2txtLoader,UnstructuredWordDocumentLoader
```

```python
### Method1: using docxtxtloader
import docx


print("Using Docxtxtloader")
try:
    docx_loader=Docx2txtLoader(r"data\word_files\resume.docx")
    docs=docx_loader.load()
    print(f"loaded {len(docs)} documents")
    print(f"Content preview: {docs[0].page_content[:200]}....")
    print(f"Metadata: {docs[0].metadata}")

except Exception as e:
    print(f"error: {e}")
```

```python
###method 2 using UnstructuredWordDocumentLoader
print("\n Using UnstructuredWordDocumentLoader")

try:
    unstructured_loader=UnstructuredWordDocumentLoader(r"data\word_files\resume.docx",mode="elements")
    unstructured_doc=unstructured_loader.load()
    print(f"Loaded {len(unstructured_doc)} elements")
    for i , doc in enumerate(unstructured_doc[:3]):
        print(f"\n Element{i+1}:")
        print(f"Type: {doc.metadata.get("category","unknown")}")
        print(f"Content: {doc.page_content[:100]}.....")


except Exception as e:
    print(e)    
```
